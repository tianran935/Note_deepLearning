# linear_classifier

## 1. 原理

### 1.1 algebraic viewpoint

$f(x,W) = Wx + b$

stretch the input matrix into the form of a vector.

1. 一个trick:

![linear classifier](linear_classifier_bias_trick_1.png)

![linear classifier](linear_classifier_bias_trick_2.png)

但是该trick在深度学习中并不常用，更倾向于分开处理权重、偏置

2. predictions are all linear


### 1.2 visual viewpoint

W矩阵，每一行是一个特征，比如我将下图拆为三个特征（3行），每一行都对应一个template，其实是在计算目标图和这个template的cosine 相似度。

![visual viewpoint](linear_classification_visual_viewpoint.png)

延伸出了linear classifier 的局限：

1. a single template cannot capture multiple modes of the data
比如，一个马的朝向可能非常不一样，朝左或朝右，但是对于linear classifier 来说，只能用一个template来表示，最后学习结果或许就是a horse with two heads.

2. 无法识别深度语义，比如对于在森林里的车，可能会和森林里的鹿非常相似。

### 1.3 geometric viewpoint

![geometric view](linear_classifier_geometric_viewpoint.png)

用于识别可以用linear classifier的情况，比如xor就是不可被识别的。

perceptron其实就是一种线性分类器。

## 2. loss function

to measure how good a W is.

loss of a single sample:

$L_i(f(x_i, W),y_i)$

loss of a whole dataset:

$L(W) = \frac{1}{N}\sum_{i=1}^N L_i(f(x_i, W),y_i)$
