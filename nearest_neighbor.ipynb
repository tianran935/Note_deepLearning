# nearest neighbor
## 原理
train: memorize all data and labels
test: predict the label of the most similar training image

### how to measure similarity?
$L_1$ distance: $d(x,y) = \sum_i |x_i - y_i|$
![example](L1_distance.png)

## 拓展
1. 训练快而推理慢，部署后无法使用，优化在这里：https://github.com/facebookresearch/faiss
2. from nearest neighbor to k-nearest neighbor：
K=1：只看离它最近的 1 个样本，它是什么类，新样本就是什么类。
K=3：看最近的 3 个样本，哪个类别多，就判成哪个类别。
K=5：看最近的 5 个样本，以此类推。
3. from L1 distance to other distance:
$L_2$ distance: $d(x,y) = \sqrt{\sum_i (x_i - y_i)^2}$
tf-idf: $d(x,y) = \sqrt{\sum_i (x_i - y_i)^2}$
4. distance metrics on pixels are not informative.  平移会带来非常大的影响
nearest neighbor with convnet

## 超参
1. K: 1, 3, 5, 10, 20, 50, 100, 200, 500, 1000
2. distance: L1, L2, tf-idf

## 调参
1. 三分法
2. cross validation (5 folds)
