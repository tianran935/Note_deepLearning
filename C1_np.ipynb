````python
import numpy as np

image = np.arange(12).reshape(2, 2, 3)
flat = image.reshape(2,6)

print(image,flat)
````

````python
scores = image
np.sort(scores)
np.argsort(scores)
print(np.argmax(scores))
print(np.argmin(scores))

````

NumPy 按轴计算常见参数：

### `axis=None`

对所有元素计算：

````python
X = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

np.sum(X)
# 21
````

### `axis=0`

消除第 0 维，对每一列计算：

````python
np.sum(X, axis=0)
# [5, 7, 9]
````

### `axis=1`

消除第 1 维，对每一行计算：

````python
np.sum(X, axis=1)
# [6, 15]
````

### `keepdims=True`

保留被消除的维度，长度设为 1：

````python
np.sum(X, axis=1, keepdims=True)
# [[ 6],
#  [15]]

# shape: (2, 1)
````

常见按轴函数：

````python
np.sum(X, axis=0)      # 求和
np.mean(X, axis=0)     # 平均值
np.max(X, axis=0)      # 最大值
np.min(X, axis=0)      # 最小值
np.std(X, axis=0)      # 标准差
np.argmax(X, axis=0)   # 最大值的索引
np.argmin(X, axis=0)   # 最小值的索引
````

记忆方法：

> `axis=n` 表示压缩掉第 `n` 个维度。

对于形状 `(2, 3)`：

```text
axis=0 → 去掉 2 → 结果形状 (3,)
axis=1 → 去掉 3 → 结果形状 (2,)
```
